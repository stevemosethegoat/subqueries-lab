# DATA ENGINEERING LIFECYCLE
# PYTHON

Data scientists often use or start data that is already available either in csv formart a .db,anAPI,dahboard but in the real production environment that data doesnt  just appear but goes through a procces for it to be available.that procces is caled the data engineering lifecycle<br>

this procces is essential to a data scientist since the quality of your data determines the quality of your analysis and  the quality of perfomance  of your models.<br>

data engineering lifecycle has the following connected steps
- data generation
- data storage
- data ingestion
- data transformation
- data serving


**Data Generation** is the point at wgich data is  captured or generated<br>
**Data Storage**  saving the data in an organised manner once it has been cleaned<br>
**Data Transformation** it is the procces of preparing raw data<br>
**Data Ingestion** procces of loading data into our system<br>
**Data serving** making data available to the end user<br>

### DATA STORAGE

```python
import sqlite3
conn= sqlite3.connect("students.db")
```

### STORAGE


```python
creating_tables="""
    CREATE TABLE IF  NOT EXISTS student_info(
        student_id interger PRIMARY KEY,
        name TEXT NOT NULL,
        phone TEXT NOT NULL,
        email TEXT UNIQUE,
        course TEXT NOT NULL,
        profile_pic BLOB
    )

"""
conn.executescript(creating_tables)
```

## INGESTION

```python
inserting_data="""
INSERT INTO student_info(name,phone,email,course,profile_pic)
VALUES
('sharleen','0713234456','sharleen@school.com','Data Science','qr.png'),
('Steve','0712434662','steve@school.com','Software Engineering','pic.png'),

('Grace','','grace@school.com','Cyber security','');

"""
conn.executescript(inserting_data)
```
